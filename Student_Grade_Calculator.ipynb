```python
marks_list = [100,84,73,37,51,66,81,70,66 ]
```

```python

for i, marks in enumerate(marks_list, start=1):

    if marks < 0 or marks > 100:
        grade = "enter  Marks between (1-100)"
    elif marks >= 90:
        grade = "A"
    elif marks >= 80:
        grade = "B"
    elif marks >= 70:
        grade = "C"
    elif marks >= 50:
        grade = "D"
    else:
        grade = "F"

    print(f"Student {i}: Marks = {marks}  got {grade} grade")
```
